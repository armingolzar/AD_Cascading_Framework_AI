### AD cascading framework research
This file is about the research on different stages of the AD cascading framework and the implementation of the AI system for 
this problem.

***

#### Stage 1
**Data:**

    In the stage 1 we have some factors like age(extremely important), biologic sex, years of education, MMSE(cognitive test which mesures memory and attention and orientation), MoCA(cognitive test which is extremely sensetive to early Mild Cognitive Impairment) and maybe some low-dimention latent space for voice and wearable gadgets that collect data from your health.

**Solution:**

    Build a simple neural network for checking the situation of the patients.
